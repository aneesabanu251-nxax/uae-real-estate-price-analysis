```python
import pandas as pd
import numpy as np
import matplotlib . pyplot as plt
import seaborn as sns
print("all libraries loaded successfully!")
```

```python
# Load the CSV file
df = pd.read_csv('uae_properties.csv')
# Always do these 3 checks first
print("Shape:", df.shape)          # rows and columns
print("\nFirst 5 rows:")
df.head()                          # preview
```

```python

print("==Data Types==")   #understand the data
print(df.dtypes)
print("==Missing Values==")
print(df.isnull().sum())
print("==Basic Statistics==")
print(df.describe)
      
      
```

```python
print("==Basic Statistics==")
print(df.describe())
```

```python
#2 Fix the date column
df['addedOn']= pd.to_datetime(df['addedOn'])

#Extract useful features from date
df['year']=df['addedOn'].dt.year
df['month']=df['addedOn'].dt.month
df['month_name']=df['addedOn'].dt.strftime('%B')

#3 Extract area from displayAddress
df['area']= df['displayAddress'].apply(lambda x: x.split(',')[-2].strip()
                                         if len(x.split(',')) >=2
                                         else 'Unknown')
                                        
 #4 Fill the missing rera values 
df['rera'] =df['rera'].fillna(0)  

#5 Drop columns not useful for analysis
df = df.drop(columns=['id','type','propertyType'])

#6 Check cleaned data
print("Cleaned Shape:",df.shape)
print("New Columns:",df.columns.tolist())
print("Missing Values after cleaning:")
print(df.isnull().sum())
```

```python
# Save cleaned data
df.to_csv('uae_properties.csv', index=False)
print("Cleaned data saved successfully")
```
